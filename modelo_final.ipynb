Este cuaderno carga `modelo_final.joblib`, genera las predicciones del conjunto de competición y guarda el archivo final `predicciones.csv`.

```python
import joblib
import pandas as pd

model = joblib.load('modelo_final.joblib')
X_comp = pd.read_pickle('bank_competition_13.pkl')

pred = model.predict(X_comp)
pred_df = pd.DataFrame({'deposit': pred})
pred_df.to_csv('predicciones.csv', index=False)

print('Predicciones guardadas en predicciones.csv')
print(pred_df['deposit'].value_counts())
pred_df.head()
```
